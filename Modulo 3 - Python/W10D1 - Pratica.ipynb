# ESERCIZIO 16


```python
guadagni = [100, 90, 70, 40, 50, 80, 90, 120, 80, 20, 50, 50]

for i in range(len(guadagni)):
    if i == 0:
        print("Mese", i+1, ":", guadagni[i], "€")
    else:
        precedenti = guadagni[:i]
        media_prec = sum(precedenti) / len(precedenti)
        if guadagni[i] > media_prec:
            print("Mese", i+1, ":", guadagni[i], "€ (media prec:", media_prec, "€ - il guadagno attuale è maggiore)")
        elif guadagni[i] < media_prec:
            print("Mese", i+1, ":", guadagni[i], "€ (media prec:", media_prec, "€ - il guadagno attuale è minore)")
        else:
            print("Mese", i+1, ":", guadagni[i], "€ (media prec:", media_prec, "€ - il guadagno attuale è uguale)")
```
