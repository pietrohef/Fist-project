# Esercizio 2


```python
# ds = DataSet 
ds = pd.read_csv("diabetes.csv")
```

```python
# 2.1 anteprima di cinque righe 
ds.head(5)
```

```python
# 2.2 metadati delle colonne
ds.info()
```

```python
# 2.3 descrittori stadistici
ds.describe()
```

```python
# 2.4 dati relativi a diverse fase di età (le query)
# si cerca il nome della columna età
ds.columns
```

```python
#query 1
ds20_29 = ds[(ds['Age (years)']>=20) & (ds['Age (years)']<=29)]
```

```python
#query 2
ds30_39 = ds[(ds['Age (years)']>=30) & (ds['Age (years)']<=39)]
```

```python
#query 3
ds40_50 = ds[(ds['Age (years)']>=40) & (ds['Age (years)']<=50)]
```

```python
# 2.5 media della pressione sanguigna diastolica per le diverse fasce di età?
media20_29 = ds20_29["Diastolic blood pressure (mm Hg)"].mean()
media30_39 = ds30_39["Diastolic blood pressure (mm Hg)"].mean()
media40_50 = ds40_50["Diastolic blood pressure (mm Hg)"].mean()

print(media20_29,media30_39,media40_50)
```

# Esercizio 4.2


```python
# 4.2.1 Pokemon leggendari
legendary = ds_pokemon[ds_pokemon["Legendary"] == True]
legendary
```

```python
# 4.2.2 Pokemon leggendari tipo 1 "grass"
legendary_grass = ds_pokemon[(ds_pokemon["Legendary"] == True) & (ds_pokemon["Type 1"] == "Grass")]
# chiave principale = ds_pokemon[(ds_pokemon["Legendary"] == True)
# chiave secondaria =  (ds_pokemon["Type 1"] == "Grass")
legendary_grass
```

```python
# 4.2.3 Pokemon leggendari tipo 1 " Ice anda Fire"
legendary_ice_and_fire= ds_pokemon[(ds_pokemon["Legendary"] == True) & ((ds_pokemon["Type 1"] == "Ice") |(ds_pokemon["Type 1"] == "Fire"))]
# chiave principale = ds_pokemon[(ds_pokemon["Legendary"] == True) 
# chiavi secondarie = ((ds_pokemon["Type 1"] == "Ice") |(ds_pokemon["Type 1"] == "Fire"))

legendary_ice_and_fire
```

```python
# 4.2.4 Pokemon prima generazione, attacco > 50 e HP < 60
prima_gen = ds_pokemon[(ds_pokemon["Generation"] == 1) & (ds_pokemon["Attack"] > 50) & (ds_pokemon["HP"] < 60)]
prima_gen
```
