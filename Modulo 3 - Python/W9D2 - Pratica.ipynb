```python
somma = 0 #inizio 
num = input("inserire un numero") #numero imput dal utente
while 0 <= somma <= 50: #condizione
    num = int(num) # cambio di tipo di dato a un intero
    somma += num 
print("La somma è", somma)
```

```python
#Possiamo usare un costrutto while per calcolare i primi 10 numeri della sequenza di Fibonacci:
numero1 = 1
numero2 = 1
fibo = [1,1] #qui si darano i risultati
conteggio = 3 
while conteggio <= 10:
    prossimo_numero = numero1 + numero2
    fibo.append(prossimo_numero)
    numero1 = numero2
    numero2 = prossimo_numero
    conteggio += 1
print(" I primi 10 numeri di Fibonacci sono:", fibo)
    
```

# ESERCIZIO 18


```python
ordini = [25, 80, 120, 15, 300, 250, 21, 78, 140]
totale_generale = 0

for ordine in ordini:
    if ordine < 100:
        spedizione = 15
    elif 100 <= ordine <= 200:
        spedizione = 5
    else:
        spedizione = 0
    
    totale_ordine = ordine + spedizione
    print("Ordine:", ordine, "=> con spedizione:", totale_ordine)
    totale_generale += totale_ordine

print("Totale di tutti gli ordini:", totale_generale)
```
